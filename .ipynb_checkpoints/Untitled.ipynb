# Customer Churn Prediction

## Objective

The objective of this project is to build a Machine Learning classification
model that predicts whether a customer will churn (`Yes`) or not churn (`No`).

The project includes:
- Data loading and understanding
- Data quality checking
- Data cleaning
- Exploratory Data Analysis
- Feature preprocessing
- Training multiple classification models
- Model evaluation and comparison
- Saving the final trained pipeline
- Reloading the saved pipeline and testing a prediction

# Prepare X and y

```python
# Remove customer_id because it is an identifier, not a useful predictive feature
X = df.drop(columns=["churn", "customer_id"])

# Target
y = df["churn"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Decision: Removing customer_id

The `customer_id` column is removed from the model features because it is
an identifier rather than a meaningful customer behavior variable.

The model should learn from customer characteristics such as age, tenure,
charges, contract type, support calls and payment method rather than from
the customer's ID.

```python
numeric_columns = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

print("Numerical columns:")
print(numeric_columns)
```

```python
categorical_columns = X.select_dtypes(include=["object"]).columns.tolist()

print("Categorical columns:")
print(categorical_columns)
```

```python
y = y.map({
    "No": 0,
    "Yes": 1
})

print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# Numerical preprocessing
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# Combine numerical and categorical preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_columns),
        ("cat", categorical_transformer, categorical_columns)
    ]
)

print("Preprocessing pipeline created.")
```

```python
logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ]
)

logistic_pipeline.fit(X_train, y_train)

print("Logistic Regression trained successfully.")
```

```python
y_pred_logistic = logistic_pipeline.predict(X_test)

print("Predictions generated.")
```

```python
accuracy_logistic = accuracy_score(y_test, y_pred_logistic)
precision_logistic = precision_score(y_test, y_pred_logistic)
recall_logistic = recall_score(y_test, y_pred_logistic)
f1_logistic = f1_score(y_test, y_pred_logistic)

print("Logistic Regression")
print("-------------------")
print("Accuracy :", accuracy_logistic *100)
print("Precision:", precision_logistic *100)
print("Recall   :", recall_logistic *100)
print("F1 Score :", f1_logistic *100)
```

```python
cm_logistic = confusion_matrix(y_test, y_pred_logistic)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_logistic,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Churn", "Churn"],
    yticklabels=["No Churn", "Churn"]
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
print("Logistic Regression Classification Report")
print("------------------------------------------")

print(
    classification_report(
        y_test,
        y_pred_logistic,
        target_names=["No Churn", "Churn"]
    )
)
```

```python
random_forest_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced"
        ))
    ]
)

random_forest_pipeline.fit(X_train, y_train)

print("Random Forest trained successfully.")
```

```python
y_pred_rf = random_forest_pipeline.predict(X_test)

print("Random Forest predictions generated.")
```

```python
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

print("Random Forest")
print("-------------")
print("Accuracy :", accuracy_rf *100)
print("Precision:", precision_rf *100)
print("Recall   :", recall_rf *100)
print("F1 Score :", f1_rf *100)
```

```python
cm_rf = confusion_matrix(y_test, y_pred_rf)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Churn", "Churn"],
    yticklabels=["No Churn", "Churn"]
)

plt.title("Random Forest - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
print("Random Forest Classification Report")
print("-----------------------------------")

print(
    classification_report(
        y_test,
        y_pred_rf,
        target_names=["No Churn", "Churn"]
    )
)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy_logistic,
        accuracy_rf
    ],
    "Precision": [
        precision_logistic,
        precision_rf
    ],
    "Recall": [
        recall_logistic,
        recall_rf
    ],
    "F1 Score": [
        f1_logistic,
        f1_rf
    ]
})

comparison
```

## Final Model Selection

After comparing Logistic Regression and Random Forest using the testing
dataset, Logistic Regression is selected as the final model.

The decision is not based only on accuracy. Precision, recall and F1-score
are also considered.

For this dataset, Logistic Regression provides a higher overall F1-score
and accuracy than Random Forest. Therefore, Logistic Regression is selected
as the final model.

The preprocessing and model will be saved together in a single pipeline
so that the same preprocessing is automatically applied when the model is
used by the Streamlit application.

```python
final_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ]
)

final_pipeline.fit(X_train, y_train)

print("Final model trained successfully.")
```

```python
joblib.dump(
    final_pipeline,
    "churn_pipeline.pkl"
)

print("Model saved as churn_pipeline.pkl")
```

```python
loaded_model = joblib.load("churn_pipeline.pkl")

print("Saved model loaded successfully.")
```

```python
# Select one customer from the testing dataset
sample_customer = X_test.iloc[[0]]

print("Customer information:")
display(sample_customer)

# Predict
prediction = loaded_model.predict(sample_customer)[0]

if prediction == 1:
    print("Predicted Churn: Yes")
else:
    print("Predicted Churn: No")
```

```python
probability = loaded_model.predict_proba(sample_customer)[0]

print("Probability of No Churn:", probability[0])
print("Probability of Churn:", probability[1])
```

```python
final_predictions = final_pipeline.predict(X_test)

print("Final Model Performance")
print("=======================")

print("Accuracy :", accuracy_score(y_test, final_predictions))
print("Precision:", precision_score(y_test, final_predictions))
print("Recall   :", recall_score(y_test, final_predictions))
print("F1 Score :", f1_score(y_test, final_predictions))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        final_predictions,
        target_names=["No Churn", "Churn"]
    )
)
```
